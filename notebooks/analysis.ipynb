```python
from pathlib import Path
# المجلد الأب للمجلد الحالي
PROJECT_ROOT = Path.cwd().parent
# الحد الأقصى للعناصر التي نعرضها داخل كل مجلد رئيسي
MAX_ITEMS = 30
# ترتيب العناصر الرئيسية حسب الاسم
items = sorted(PROJECT_ROOT.iterdir(), key=lambda x: x.name.lower())
for item in items:
    if item.is_dir():
        print(f"📁 {item.name}/")
        subitems = sorted(item.iterdir(), key=lambda x: x.name.lower())
        if len(subitems) > MAX_ITEMS:
            for subitem in subitems[:30]:
                icon = "📁" if subitem.is_dir() else "📄"
                print(f"   ├── {icon} {subitem.name}")
            print("   ├── ...")
            # عرض آخر عنصر ضمن القائمة
            subitem = subitems[-1]
            icon = "📁" if subitem.is_dir() else "📄"
            print(f"   └── {icon} {subitem.name}")
        else:
            # إذا كان عدد العناصر 10 أو أقل
            for i, subitem in enumerate(subitems):
                icon = "📁" if subitem.is_dir() else "📄"
                # تحديد شكل الخط الأخير
                connector = "└──" if i == len(subitems) - 1 else "├──"
                print(f"   {connector} {icon} {subitem.name}")
    else:
        print(f"📄 {item.name}")
```

# CSV to PostgreSQL Tables (before jupyter)

```python
import pandas as pd
customers = pd.read_csv("../data/olist_customers_dataset.csv")
customers.head()
```

```python
from pathlib import Path

DATA_DIR = Path("../data")
files = list(DATA_DIR.glob("*.csv"))

for f in files:
    print(f.name)
```

```python
datasets = {}

for file in files:
    name = file.stem.replace("olist_", "").replace("_dataset", "")
    datasets[name] = pd.read_csv(file)
    print(f"{name:35} {datasets[name].shape}")
```

```python

table_names = {
    "customers": "customers",
    "geolocation": "geolocation",
    "orders": "orders",
    "order_items": "order_items",
    "order_payments": "order_payments",
    "order_reviews": "order_reviews",
    "products": "products",
    "sellers": "sellers",
    "product_category_name_translation": "category_translation"
}
for key, table in table_names.items():
    datasets[key].to_sql(
        table,
        engine,
        if_exists="replace",
        index=False,
        chunksize=5000
    )
```
